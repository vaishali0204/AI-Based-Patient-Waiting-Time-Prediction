```python
features = [
    "Patient Type",
    "Queue Count Before Processing",
    "Arrival Hour",
    "Arrival Minute",
    "Time Period"
]

target = "Total Queue Waiting (min)"

X = model_data[features].copy()
y = model_data[target].copy()

print("X shape:", X.shape)
print("y shape:", y.shape)

display(X.head())
display(y.head())
```

```python
X.info()
```

```python
print(X.isnull().sum())
print(y.isnull().sum())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

```python
categorical_features = [
    "Patient Type",
    "Time Period"
]

numerical_features = [
    "Queue Count Before Processing",
    "Arrival Hour",
    "Arrival Minute"
]
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

preprocessor = ColumnTransformer(
    transformers=[
        (
            "categorical",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_features
        ),
        (
            "numerical",
            "passthrough",
            numerical_features
        )
    ]
)
```

```python
from sklearn.pipeline import Pipeline
from sklearn.linear_model import Ridge

ridge_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("model", Ridge(alpha=1.0))
    ]
)
```

```python
ridge_model.fit(X_train, y_train)

print("Ridge model trained successfully.")
```

```python
y_pred = ridge_model.predict(X_test)

print("First 10 predictions:")
print(y_pred[:10])
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

mae = mean_absolute_error(y_test, y_pred)

rmse = np.sqrt(
    mean_squared_error(y_test, y_pred)
)

r2 = r2_score(y_test, y_pred)

print("Ridge Regression Results")
print("------------------------")
print(f"MAE  : {mae:.2f} minutes")
print(f"RMSE : {rmse:.2f} minutes")
print(f"R²   : {r2:.4f}")
```

```python
comparison = pd.DataFrame({
    "Actual Waiting Time": y_test.values,
    "Predicted Waiting Time": y_pred
})

comparison["Error"] = (
    comparison["Actual Waiting Time"]
    - comparison["Predicted Waiting Time"]
)

display(comparison.head(10))
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(7, 5))

plt.scatter(y_test, y_pred)

plt.xlabel("Actual Waiting Time (minutes)")
plt.ylabel("Predicted Waiting Time (minutes)")
plt.title("Ridge Regression: Actual vs Predicted")

# Perfect prediction line
plt.plot(
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()],
    linestyle="--"
)

plt.show()
```

```python
from sklearn.ensemble import RandomForestRegressor

rf_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        (
            "model",
            RandomForestRegressor(
                n_estimators=200,
                random_state=42,
                max_depth=None,
                min_samples_split=2
            )
        )
    ]
)
```

```python
rf_model.fit(X_train, y_train)

print("Random Forest model trained successfully.")
```

```python
rf_pred = rf_model.predict(X_test)

print("First 10 Random Forest predictions:")
print(rf_pred[:10])
```

```python
rf_mae = mean_absolute_error(y_test, rf_pred)

rf_rmse = np.sqrt(
    mean_squared_error(y_test, rf_pred)
)

rf_r2 = r2_score(y_test, rf_pred)

print("Random Forest Results")
print("--------------------")
print(f"MAE  : {rf_mae:.2f} minutes")
print(f"RMSE : {rf_rmse:.2f} minutes")
print(f"R²   : {rf_r2:.4f}")
```
